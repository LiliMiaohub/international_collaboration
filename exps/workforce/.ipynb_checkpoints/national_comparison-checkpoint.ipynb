this script compares the scientific workforce of countries in international collaboration and domestically. the way of testing it is to bootstrapping researchers from each country and compare the expected results with the actual international workforce.

```python
import pandas as pd
from time import time
import numpy as np
import tqdm
import geopandas as gp
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.axes_divider import make_axes_locatable
import seaborn as sns
```

```python
pub_detail_path='../../data/dropbox/LaborDivision/Data/deriveddata/pub_detail.pkl'
author_path='../../data/dropbox/LaborDivision/Data/deriveddata/authors_ordered.pkl'
author_info_path='../../data/dropbox/LaborDivision/Data/cleanedraw/author_info.pkl'
ingroup_path = '../../data/dropbox/LaborDivision/Data/additionaldata/cntry_meta.csv'
```

```python
pub_all=pd.read_pickle(pub_detail_path)
```

```python
pub_int=pub_all[pub_all.int==1]
pub_int.head()
```

```python
author_all=pd.read_pickle(author_path)
author_all.head()
```

```python
author_info=pd.read_pickle(author_info_path)
author_info.head()
```

```python
author_all=author_all.merge(author_info,on=['id','seqno'],how='left')
```

```python
author_int=author_all[author_all['id'].isin(pub_int['id'].tolist())]
```

```python
author_int=author_int[['cluster_id','Code','gender','yfp','npub']]
author_int=author_int.drop_duplicates(subset=['cluster_id','Code'])
```

```python
author_int.head()
```

```python
author_all=author_all[['cluster_id','Code','gender','yfp','npub']]
author_all=author_all.drop_duplicates(subset=['cluster_id','Code'])
```

```python
#number of authors in that country are participated in international collaboration
cntry_nint=author_int.groupby(['Code']).size().reset_index(name='count')
cntry_nint=dict(zip(cntry_nint.Code.values,cntry_nint['count'].values))
```

```python
N=10 #number of bootstrapping
sample_res=[]
for cntry,num in tqdm.tqdm(cntry_nint.items()):
    n=num#size to sample which is the actual number of international authors
    pop_df = author_all[author_all.Code==cntry] #population to sample which is the number of all authors in the cntry
    sample_df=pop_df.sample(n=n*N, replace=True)
    try:
        nf=sample_df['gender'].value_counts().F
    except AttributeError:
        nf=0
    try:
        nm=sample_df['gender'].value_counts().M
    except AttributeError:
        nm=0
    npub=sample_df['npub'].mean()
    nyfp=sample_df['yfp'].mean()
    sample_res.append([cntry,nf,nm,npub,nyfp])
```

```python
sample_df=pd.DataFrame(sample_res,columns=['Code','nf','nm','npub','yfp'])
```

```python
sample_df.head()
sample_df['nf']=sample_df['nf']/10
sample_df['nm']=sample_df['nm']/10
```

```python
part1 = author_int.groupby('Code').agg({'yfp':'mean','npub':'mean'}).reset_index()
part1.head()
```

```python
part2 = author_int.groupby(['Code','gender']).size().reset_index(name='count')
part2=part2.pivot_table(index='Code',columns='gender',values='count',fill_value=0).reset_index()
part2=part2[['Code','F','M']]
part2=part2.rename(columns={'F':'nf','M':'nm'})
```

```python
act_df=part1.merge(part2,on='Code')
```

```python
act_df.head()
```

```python
meta=act_df.merge(sample_df,on='Code',suffixes=('_r', '_s'))
```

```python
meta['yfp']=meta['yfp_r']-meta['yfp_s']
meta['npub']=(meta['npub_r']-meta['npub_s'])/meta['npub_s']

meta['nf']=(meta['nf_r']-meta['nf_s'])/meta['nf_s']
meta['nm']=(meta['nm_r']-meta['nm_s'])/meta['nm_s']
```

```python
ingroup_df=pd.read_csv(ingroup_path)[['Code','ingroup_08-20']]
ingroup_df=ingroup_df.drop_duplicates()
ingroup_df.head()
```

```python
meta=meta.merge(ingroup_df,on='Code')
```

```python
meta.head()
```

```python
world_geo=gp.read_file("../../data/dropbox/LaborDivision/Data/additionaldata/worldmap/ne_50m_admin_0_countries.shp")
```

```python
world_geo = world_geo[['ADMIN','NAME','ISO_A3','geometry']]
world_geo=world_geo[world_geo.NAME!='Antarctica']
```

```python
def plot_worldmap(ax, fig, basemap, data, plotcolumn, cmap, vmin, vmax,legend_title):
    """plot a basemap and color the country based on the column passed into
    """

    basemap.plot(ax=ax, linewidth=0.25, edgecolor="#8C8C8C", facecolor="#ffffff")
    data.plot(column=plotcolumn, cmap=cmap, vmin=vmin, vmax=vmax, ax=ax)
    ax.axis("off")
    
    divider = make_axes_locatable(ax)
    cax = divider.append_axes("right", size="2%", pad=0.05)
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=vmin, vmax=vmax))
    sm._A = []
    clb = fig.colorbar(sm, cax=cax)
    clb.ax.set_title(legend_title)

    return fig, ax
```

```python
meta.head()
```

```python
meta['npub'].min()
```

```python
fig, ax=plt.subplots(nrows=1,ncols=2,figsize=(20,6),gridspec_kw={'width_ratios': [4, 2]})
plt.subplots_adjust(wspace=0.12,hspace=0.1)
df = meta.merge(world_geo,left_on='Code',right_on='ISO_A3')
df = gp.GeoDataFrame(df)
plot_worldmap(ax[0],fig,world_geo,df,'yfp','viridis',-3,1,'yfp')
sns.kdeplot(x='yfp',hue='ingroup_08-20',data=meta,ax=ax[1],hue_order=['H','UM','LM','L'])
```

```python
fig, ax=plt.subplots(nrows=1,ncols=2,figsize=(20,6),gridspec_kw={'width_ratios': [4, 2]})
plt.subplots_adjust(wspace=0.12,hspace=0.1)
plot_worldmap(ax[0],fig,world_geo,df,'npub','viridis',-0.3,1.8,'npub')
sns.kdeplot(x='npub',hue='ingroup_08-20',data=meta,ax=ax[1],hue_order=['H','UM','LM','L'])
```

```python
fig, ax=plt.subplots(nrows=1,ncols=2,figsize=(20,6),gridspec_kw={'width_ratios': [4, 2]})
plt.subplots_adjust(wspace=0.12,hspace=0.1)
plot_worldmap(ax[0],fig,world_geo,df,'nf','viridis',-0.35,0.20,'female')
sns.kdeplot(x='nf',hue='ingroup_08-20',data=meta,ax=ax[1],hue_order=['H','UM','LM','L'])
```
