this script reclassifies countries based on rules

```python
import pandas as pd
```

```python
income_path="../../data/dropbox/LaborDivision/Data/additionaldata/cntry_meta.csv"
```

```python
cntry_group_path='../../data/dropbox/LaborDivision/Data/additionaldata/cntry_classification/cntry_group_meta.csv'
```

```python
income_df=pd.read_csv(income_path)
income_df.head()
```

```python
top20_nature=['USA','CHN','DEU','GBR','JPN','FRA','CAN','KOR','CHE','AUS',
              'IND','ITA','ESP','NLD','SWE','ISR','SGP','RUS','DNK','BEL']
```

```python
group_df=income_df[['Code','ingroup_08-20']].drop_duplicates()
group_df['income_ni']='L'
group_df.loc[
    (group_df['Code'].isin(top20_nature))&(group_df['ingroup_08-20']=='H'), 'income_ni'] = 'H'
```

```python
group_df.to_csv(cntry_group_path,index=False)
```
