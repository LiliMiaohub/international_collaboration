```python
cntry_dis_dict = {}
cntry_dis_pubs_array = cntry_dis_pubs[['cntry','dis','intl','uid']].values
# Iterate through the dataframe rows
for row in tqdm(cntry_dis_pubs_array):
    cntry = row[0]
    dis = row[1]
    intl = row[2]
    uid = row[3]
    
    # Check if the key_A already exists in the nested dictionary
    if (cntry,dis) in cntry_dis_dict:
        # Check if the key_B already exists for the key_A
        if intl in cntry_dis_dict[(cntry,dis)]:
            # Append the value_C to the existing list
            cntry_dis_dict[(cntry,dis)][intl].append(uid)
        else:
            # Create a new key_B entry with a list containing value_C
            cntry_dis_dict[(cntry,dis)][intl] = [uid]
    else:
        # Create a new key_A entry with a nested dictionary containing key_B and value_C
        cntry_dis_dict[(cntry,dis)] = {intl: [uid]}
```

```python
def return_embeddings(pubs,uid_index_dict,embedding):
    embed_index = [uid_index_dict[pub] for pub in pubs if pub in uid_index_dict]
    if len(embed_index)>0:
        embed_vectors = embedding[embed_index]
        return embed_vectors
    else:
        return np.array([])
```

```python
def resample_data(array, sample_minimum=20, sample_max=10000):
    if len(array) < sample_minimum:
        resampled_indices = np.random.choice(len(array), size=sample_minimum, replace=True)
        return array[resampled_indices]
    #elif len(array) > sample_max:
        #resampled_indices = np.random.choice(len(array), size=sample_max, replace=True)
        #return array[resampled_indices]
    return array     
```

```python
def compare_pairwise_similarity(u_values, v_values):
    
    resampled_u = resample_data(u_values)
    resampled_v = resample_data(v_values)
    try:
        similarity = cosine_similarity(resampled_u, resampled_v)
    except:
        print(u_values)
        print(v_values)
    
    return np.mean(similarity)   
```

```python
def process_similarity(args):
    cntry_dis, dis_dict, uid_index_dict, embedding = args
    res = []
    for cntry_dis, dis_dict in tqdm(cntry_dis_dict.items()):
        if 0 in dis_dict and 1 in dis_dict:
            dome_embeddings = return_embeddings(dis_dict[0], uid_index_dict, embedding)
            intl_embeddings = return_embeddings(dis_dict[1], uid_index_dict, embedding)

            if (dome_embeddings.shape[0] > 5 and intl_embeddings.shape[0] > 5):

                cosine = compare_pairwise_similarity(dome_embeddings, intl_embeddings)
                res.append([cntry_dis[0], cntry_dis[1], 'cosine', cosine])
    return res
```

```python
def compute_similarity(cntry_dis_dict, uid_index_dict, embedding):
    args = [(cntry_dis, dis_dict, uid_index_dict, embedding) for cntry_dis, dis_dict in cntry_dis_dict.items()]
    
    with Pool() as pool:
        results = pool.map(process_similarity, args)
    res = [item for sublist in results for item in sublist]
    return res
```

```python
res = compute_similarity(cntry_dis_dict, uid_index_dict, embedding)
```

```python
cntry_dis_simi = pd.DataFrame(res, columns=['cntry','dis','type','similarity'])
cntry_dis_simi.head()
```

### bootstrapping the results

```python
cntry_simi = cntry_simi[cntry_simi.type=='cosine']
sts = cntry_simi.ST.unique()
sample_size = cntry_simi[cntry_simi.ST=='Advanced'].shape[0]
bootstrapping_res = []

for st in sts:
    for i in range(20):
        df = cntry_simi[cntry_simi.ST==st]
        df_sampled = df.sample(n=sample_size, replace=True)
        similarity_sampled = df_sampled['similarity'].mean()
        bootstrapping_res.append([similarity_sampled, st])
                          
```

```python
btsp_res = pd.DataFrame(bootstrapping_res, columns=['simi','st'])
```

```python
fig, ax=plt.subplots()
color={'Advanced':'#E01A27','Proficient':'#F88A4D','Developing':'#5DC97C','Lagging':'#23A5D9'}
order = ['Advanced','Proficient','Developing','Lagging']
sns.stripplot(x='st',y='simi',data=btsp_res,ax=ax, order=order, palette=color)
sns.boxplot(x='st',y='simi',data=btsp_res,
            boxprops={'facecolor':'none', 'edgecolor':'#A7A7A7'},
            medianprops={'color':'#A7A7A7'},
            whiskerprops={'color':'#A7A7A7'},
            capprops={'color':'#A7A7A7'},
            order=order,
            fliersize=0,
            ax=ax)
ax.set_ylabel("Similarity from Bootstrapping", fontsize=12)
ax.set_xlabel("")
ax.set_ylim(0.2,1)
ax.tick_params(axis='both', which='major', labelsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.savefig("../../data/dropbox/LaborDivision/figs/bootstrapping_pub_similarity_derived_cntry_dis.pdf",bbox_inches='tight')
```

```python
def cal_angle_dist(v1,v2):
    def dotproduct(v1, v2):
        return sum((a*b) for a, b in zip(v1, v2))
    
    def length(v):
        return math.sqrt(dotproduct(v, v))
    try:
        return math.acos(dotproduct(v1, v2) / (length(v1) * length(v2)))
    except ValueError:
        return np.zeros(64)
```
