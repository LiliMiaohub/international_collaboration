### clean the title data to convert the raw data into a table with ost_bk, paper_id and title

```python
import pandas as pd
```

```python
raw_title_path = "../../data/dropbox/LaborDivision/Data/rawdata/article_title_lb.txt"
clean_title_path = "../../data/dropbox/LaborDivision/Data/cleanedraw/pub_title.csv"
```

```python
raw_title = pd.read_csv(raw_title_path, sep="|", on_bad_lines='skip')
raw_title.head()
```

```python
raw_title = raw_title.reset_index()
```

```python
raw_title.columns=['paper_id','ost_bk','title']
raw_title.to_csv(clean_title_path, index=False)
```
