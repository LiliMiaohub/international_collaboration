### this script normalizes the raw matrix of occurance of role of each country with the expected value

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
author_path="../../data/dropbox/LaborDivision/Data/deriveddata/pubs_noalpha_int/authors_noalpha_int.pkl"
authorship_shuffle_path="../../data/dropbox/LaborDivision/Data/deriveddata/authorship_shuffle/authorship_shuffle.pkl"
cntry_meta_path = '../../data/dropbox/LaborDivision/Data/additionaldata/cntry_meta.csv'
```

```python
authorship_norm_path="../../data/dropbox/LaborDivision/Data/deriveddata/authorship_shuffle/authorship_norm.pkl"
```

```python
author_df=pd.read_pickle(author_path)
```

```python
shuffle_df=pd.read_pickle(authorship_shuffle_path)
```

```python
cntry_meta=pd.read_csv(cntry_meta_path)
```

```python
#calculate the occurrance the specific role of a country of randomized over years
shuffle_cnt=shuffle_df.groupby(['cntry','year','role']).size().reset_index(name='shuffle')
```

```python
shuffle_cnt.head()
```

```python
#calculate the occurrance of raw matrix
author_cnt=author_df.groupby(['Code','year','role']).size().reset_index(name='raw')
author_cnt=author_cnt.rename(columns={'Code':'cntry'})
author_cnt.head()
```

```python
meta_cnt=author_cnt.merge(shuffle_cnt, on=['cntry','year','role'],how='left')
meta_cnt=meta_cnt.fillna(value=0)
```

```python
n=shuffle_cnt['shuffle'].sum()/author_cnt['raw'].sum()#the number of times of the shuffling
meta_cnt['shuffle']=meta_cnt['shuffle']/n #normalize the shuffling by the number of times
```

```python
meta_cnt.to_pickle(authorship_norm_path)
```
