this script selects no alphabetically ordered and international collaborated publications for the main analysis

```python
import pandas as pd
```

```python
pub_detail_path='../../data/dropbox/LaborDivision/Data/deriveddata/pub_detail.pkl'
author_path='../../data/dropbox/LaborDivision/Data/deriveddata/authors_ordered.pkl'
```

```python
authornew_path="../../data/dropbox/LaborDivision/Data/deriveddata/pubs_noalpha_int/authors_noalpha_int.pkl"
pubnew_path="../../data/dropbox/LaborDivision/Data/deriveddata/pubs_noalpha_int/pubs_noalpha_int.pkl"
```

```python
pub_df=pd.read_pickle(pub_detail_path)
```

```python
author_df=pd.read_pickle(author_path)
```

```python
pub_df=pub_df[(pub_df.alpha=='no')&(pub_df.teamsize>2)&(pub_df.int==1)]
author_df=author_df[author_df['id'].isin(pub_df['id'].tolist())]
```

```python
author_df.to_pickle(authornew_path)
pub_df.drop(columns=['alpha','teamsize','int']).to_pickle(pubnew_path)
```
