this script detect alphabetical ordered papers and remove them from analysis. We only select internationally authored; teamsize is larger than 2 and no alphabetically ordered pubs in our analysis. 

The output contains three files:

1. author_noalpha and article_noalpha are the files that contains the above threshold pubs

2. pub_info contains the **year**, **discipline**, **list of last names**, **list of countries**, **alphabetical**, **international** information of publications

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
author_path="../../data/dropbox/LaborDivision/Data/cleanedraw/authors.pkl"
dis_path = '../../data/dropbox/LaborDivision/Data/rawdata/discipline_code.csv'
article_new='../../data/dropbox/LaborDivision/Data/cleanedraw/pubs.pkl'
```

```python
pub_detail_path='../../data/dropbox/LaborDivision/Data/deriveddata/pub_detail.pkl'
```

```python
author_df=pd.read_pickle(author_path)
```

```python
pub_df=pd.read_pickle(article_new)
```

```python
author_df.head()
```

```python
alpha_df = author_df.groupby(['id']).agg({'last':list,
                                       'Code':list})
```

```python
def alpha_order(namelist):
    nlist=namelist[:6]
    if len(nlist)==1:
        return "null"
    elif all(nlist[i] <= nlist[i+1] for i in range(len(nlist) - 1)):
        return "yes"
    else:
        return "no"
```

```python
alpha_df['alpha']=alpha_df['last'].apply(lambda x:alpha_order(x))
```

```python
# number of authors in paper. Only choose pubs with more than 3 authors into analysis
alpha_df['teamsize']=alpha_df['last'].apply(lambda x:len(x) if len(x)<4 else 4)
```

```python
# number of unique countries. the foci papers are international collaboration
alpha_df['int']=alpha_df['Code'].apply(lambda x:1 if len(set(x))>1 else 0)
```

```python
alpha_df=alpha_df.drop(columns=['last','Code'])
```

```python
pub_detail=pub_df.merge(alpha_df.reset_index(),on='id')
```

```python
pub_detail.to_pickle(pub_detail_path)
```

```python
tt1 = pub_detail[(pub_detail.int==1)&(pub_detail.alpha=='no')
              ].groupby(['teamsize']).size().reset_index(name='value')
```

```python
tt2 = pub_detail[(pub_detail.int==1)&(pub_detail.alpha=='yes')
              ].groupby(['teamsize']).size().reset_index(name='value')
```

```python
tt=pub_detail[pub_detail.int==1].shape[0]
tt1['p']=tt1['value']/tt
tt2['p']=tt2['value']/tt
```

```python
labels=['two','three','four']
noalpha=tt1['p'].values
alpha=tt2['p'].values
width = 0.35

fig, ax = plt.subplots()
ax.bar(labels, noalpha, width, label='no-alpha')
ax.bar(labels,alpha, width, bottom = noalpha,label='alpha')
plt.text(0.2,0.01,'137921(3%)')
plt.text(1.2,0.01,'421487(10%)')
plt.text(1.2,0.5,'3081268(74%)')
plt.ylabel('proportion of international collaboration')
plt.legend()
plt.savefig("../../data/dropbox/LaborDivision/figs/Data/noalph_proportion.pdf")
```
