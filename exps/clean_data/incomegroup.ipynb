this script cleans the income group data from world bank, and convert the annual income group to time period

```python
import pandas as pd
import numpy as np
```

```python
income_path='../../data/dropbox/LaborDivision/data/additionaldata/incomegroup_wb.csv'
cntry_path = '../../data/dropbox/LaborDivision/Data/additionaldata/cntry_convert.csv'
```

```python
income_df=pd.read_csv(income_path)
income_df = income_df.set_index('Code')
```

```python
years = np.arange(2008,2021)
years = [str(a) for a in years]
income_df=income_df[years]
```

```python
income_df = income_df.mode(axis=1).reset_index()
income_df.columns=['Code','ingroup']
```

```python
cntry_df=pd.read_csv(cntry_path)
cntry_df=cntry_df.dropna()
```
