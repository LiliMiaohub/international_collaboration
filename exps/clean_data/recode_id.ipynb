## this script recodes the UID to a string start from 0

```python
import pandas as pd
```

```python
article_path = '../../data/dropbox/LaborDivision/Data/rawdata/article_lb.txt'
cntry_path = '../../data/dropbox/LaborDivision/Data/rawdata/article_cntry_lb.txt'
fund_path = '../../data/dropbox/LaborDivision/Data/rawdata/article_fundagency_lb.txt'
id_map_path = '../../data/dropbox/LaborDivision/Data/cleanedraw/id_map.csv'
```

```python
def recode_id(old_id, id_map):
    if old_id in id_map:
        return id_map[old_id]
    else:
        new_id=len(id_map)
        id_map['old_id']=new_id
        return new_id
```

```python
file_names=[article_path, cntry_path, fund_path]
id_map = {}
```

```python
for file_name in file_names:
    df = pd.read_csv(article_path,sep='|',on_bad_lines='skip')
    df['pid'] = df['UID'].apply(lambda x: recode_id(x, id_map))
    
    new_file_name = file_name.replace('rawdata','cleanedraw')
    df.to_csv(new_file_name, index=False)
```

```python
with open(id_map_path, 'w') as file:
    for key, value in id_map.items():
        file.write(f"{key}:{value}\n")
```
