### this script rank country based on the nature index

```python
import pandas as pd
```

```python
raw_nature_path = '../../data/dropbox/LaborDivision/data/additionaldata/cntry_classification/Nature-Index/2023-NatureIndex_Raw.csv'
```

```python
df = pd.read_csv(raw_nature_path)
df.head()
```

```python
name2code_path = "../../data/dropbox/LaborDivision/data/additionaldata/cntry_classification/Nature-Index/cntry2code.csv"
```

```python
name2code_df = pd.read_csv(name2code_path,names=['cntry','code'])
name2code_df.head()
```

```python
df = df.merge(name2code_df, left_on='Country/territory', right_on='cntry')
```

```python
df['group']=pd.cut(df.index, bins=4, labels=['H','UM','LM','L'])
```

```python
clean_nature_path = '../../data/dropbox/LaborDivision/data/additionaldata/cntry_classification/Nature-Index/index_cleaned.csv'
df[['code','cntry','Count','Share','group']].to_csv(clean_nature_path, index=False)
```
