```python
import pandas as pd
```

```python
article_path = '../../data/dropbox/LaborDivision/Data/rawdata/articles.txt'
corr_path = '../../data/dropbox/LaborDivision/Data/rawdata/art_cntry_corr.txt'
```

```python
pub_df=pd.read_csv(article_path,sep="\t",header=0,
                   names=['id','year','dis'],
                   error_bad_lines=False, warn_bad_lines=False
                  )
corr_df=pd.read_csv(corr_path,sep="\t",header=0,
                    names=['id','full','seqno','addrno','cntry'],
                    error_bad_lines=False,warn_bad_lines=False)
```

```python
corr_df.head()
```

```python
pub_df.head()
```

```python
df=corr_df[['id','seqno']].merge(pub_df[['id','year']],on='id')
```

```python
df['label']=df.apply(lambda x: 'y' if x.seqno==1 else 'n',axis=1)
```

```python
df = df.groupby(['year','label']).size().reset_index()
```

```python
df=df.rename(columns={0:'count'})
```

```python
df = df.pivot(index='year',columns='label',values='count')
```

```python
df['total']=df['n']+df['y']
```

```python
df['p']=df['y']/df['total']
```
